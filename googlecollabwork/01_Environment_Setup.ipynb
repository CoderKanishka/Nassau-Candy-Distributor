# S Initial Data Validation

## Objective

Before performing any analysis, we must verify the quality of the dataset.

This includes:

- Dataset dimensions
- Column names
- Data types
- Missing values
- Duplicate records
- Statistical summary
- Memory usage

This step ensures that the dataset is reliable for further analysis.

```python

# ==========================================
# Dataset Shape
rows, columns = df.shape

print("="*50)
print("📊 DATASET INFORMATION")
print("="*50)

print(f"Total Rows    : {rows}")
print(f"Total Columns : {columns}")

print("="*50)
```

```python
# ==========================================
# Column Names
# ==========================================

print("Columns Present in Dataset:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# ==========================================
#  Data Types
# ==========================================

datatype_df = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values
})

datatype_df
```

```python
# ==========================================
#  Missing Values
# ==========================================

missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values
})

missing_df
```

```python
# ==========================================
#  Unique Values
# ==========================================

unique_df = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique() for col in df.columns]
})

unique_df
```

```python
# ==========================================
#  Memory Usage
# ==========================================

memory = df.memory_usage(deep=True) / 1024**2

memory_df = pd.DataFrame({
    "Column": memory.index,
    "Memory(MB)": memory.values
})

memory_df
```

```python
# ==========================================
#  Numerical Summary
# ==========================================

df.describe().T
```

```python
# ==========================================
#  Categorical Summary
# ==========================================

df.describe(include="object").T
```

```python
# ==========================================
#  Data Quality Report
# ==========================================

quality_report = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

quality_report
```

```python
# ==========================================
#  Save Report
# ==========================================

report_path = "/content/drive/MyDrive/Nassau_Candy_Project/reports/Data_Quality_Report.csv"

quality_report.to_csv(report_path, index=False)

print("✅ Data Quality Report Saved Successfully")
print(report_path)
```
