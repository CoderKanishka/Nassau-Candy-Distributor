```python
# 1. Import Libraries
import pandas as pd
import numpy as np
import os

# 2. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 3. Load Dataset
df = pd.read_csv("/content/drive/MyDrive/Nassau_Candy_Project/data/raw/Nassau Candy Distributor.csv")
```

```python
print(df.columns.tolist())
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("/", "_")
)

print(df.columns)
```

```python
for col in df.columns:
    print(col)
```

```python
# ==========================================
#  Convert Date Columns
# ==========================================

date_columns = ["order_date", "ship_date"]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("✅ Date Columns Converted Successfully")

df[date_columns].head()
```

```python
# ==========================================
# Missing Values Report
# ==========================================

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Percentage": (
        df.isnull().sum().values / len(df) * 100
    ).round(2)
})

missing_report
```

```python
negative_sales = df[df["sales"] < 0]

print("Negative Sales Records :", len(negative_sales))
```

```python
negative_cost = df[df["cost"] < 0]

print("Negative Cost Records :", len(negative_cost))
```

```python
negative_profit = df[df["gross_profit"] < 0]

print("Negative Profit Records :", len(negative_profit))
```

```python
zero_sales = df[df["sales"] == 0]

print("Zero Sales Records :", len(zero_sales))
```

```python
zero_units = df[df["units"] == 0]

print("Zero Units :", len(zero_units))
```

```python
df["calculated_profit"] = df["sales"] - df["cost"]

df["difference"] = (
    df["gross_profit"] -
    df["calculated_profit"]
).round(2)

df[[
    "sales",
    "cost",
    "gross_profit",
    "calculated_profit",
    "difference"
]].head()
```

```python
wrong_profit = df[df["difference"] != 0]

print("Incorrect Profit Records :", len(wrong_profit))
```

```python
print("="*60)

print("DATA VALIDATION SUMMARY")

print("="*60)

print("Rows :", len(df))

print("Columns :", len(df.columns))

print("Missing Values :", df.isnull().sum().sum())

print("Duplicate Rows :", df.duplicated().sum())

print("Negative Sales :", len(negative_sales))

print("Negative Cost :", len(negative_cost))

print("Negative Profit :", len(negative_profit))

print("Zero Sales :", len(zero_sales))

print("Zero Units :", len(zero_units))

print("="*60)
```

```python
validation_report = pd.DataFrame({
    "Metric":[
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows",
        "Negative Sales",
        "Negative Cost",
        "Negative Profit",
        "Zero Sales",
        "Zero Units"
    ],
    "Value":[
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        len(negative_sales),
        len(negative_cost),
        len(negative_profit),
        len(zero_sales),
        len(zero_units)
    ]
})

validation_report.to_csv(
"/content/drive/MyDrive/Nassau_Candy_Project/reports/Data_Validation_Report.csv",
index=False
)

print("✅ Validation Report Saved")
```
