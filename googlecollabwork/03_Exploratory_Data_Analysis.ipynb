```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv(
    "/content/drive/MyDrive/Nassau_Candy_Project/data/processed/processed_sales.csv"
)

df.head()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

df[["Order Date","Ship Date","Shipping Days"]].head()
```

```python
df["Profit Margin %"] = (
    df["Gross Profit"] / df["Sales"]
) * 100

df["Profit Margin %"] = df["Profit Margin %"].round(2)

df[["Sales","Gross Profit","Profit Margin %"]].head()
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
total_sales = df["Sales"].sum()

total_profit = df["Gross Profit"].sum()

total_cost = df["Cost"].sum()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

average_margin = df["Profit Margin %"].mean()

print("Total Sales :", total_sales)

print("Total Profit :", total_profit)

print("Total Cost :", total_cost)

print("Total Orders :", total_orders)

print("Total Customers :", total_customers)

print("Average Margin :", round(average_margin,2),"%")
```

```python
monthly_sales = (
    df.groupby("Month Name")["Sales"]
      .sum()
)

monthly_sales
```

```python
monthly_sales.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Monthly Sales")

plt.ylabel("Sales")

plt.show()
```

```python
top_states = (
    df.groupby("State/Province")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_states
```

```python
division_profit = (
    df.groupby("Division")["Gross Profit"]
      .sum()
      .sort_values(ascending=False)
)

division_profit
```

```python
region_profit = (
    df.groupby("Region")["Gross Profit"]
      .sum()
      .sort_values(ascending=False)
)

region_profit
```

```python
category = (
    df.groupby("Division")
      .agg(
          Sales=("Sales","sum"),
          Profit=("Gross Profit","sum"),
          Cost=("Cost","sum"),
          Units=("Units","sum")
      )
)

category
```

```python
abc = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

abc["Cumulative Sales"] = abc["Sales"].cumsum()

abc["Cumulative %"] = (
    abc["Cumulative Sales"] /
    abc["Sales"].sum()
) * 100
```

```python
def abc_category(x):

    if x <= 80:
        return "A"

    elif x <=95:
        return "B"

    else:
        return "C"

abc["ABC Category"] = abc["Cumulative %"].apply(abc_category)

abc.head()
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    abc["Product Name"],
    abc["Sales"]
)

plt.plot(
    abc["Cumulative %"],
    color="red"
)

plt.xticks(rotation=90)

plt.title("Pareto Analysis")

plt.show()
```

```python
monthly_growth = (
    df.groupby(["Year","Month"])["Sales"]
      .sum()
      .reset_index()
)

monthly_growth["Growth %"] = (
    monthly_growth["Sales"]
    .pct_change()*100
)

monthly_growth
```
